```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, accuracy_score, f1_score, confusion_matrix
from sklearn.linear_model import LogisticRegression
```

```python
data = pd.read_csv('insclass_train.csv').drop(columns = ['variable_8', 'variable_14'])
data.head()
```

```python
y = data['target']
x = pd.get_dummies(data.drop(columns = ['target']).fillna(-100))
x.head()
```

```python
x = ((x - x.mean())/x.std()).drop(columns =['variable_3'])
x.head()
```

```python
logreg = LogisticRegression(max_iter = 1500)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.1)
logreg.fit(x_train, y_train)

print(confusion_matrix(y_train, logreg.predict(x_train)))
```

$$Accuracy = \frac{TP+TN}{TP+TN+FP+FN}$$
Доля правильных ответов

```python
# Positive - плохие клиенты, Negative - хорошие
TN = 123295 # True Negative - хорошие клиенты, которых машина назвала хорошими
TP = 1893
FN = 11054 # False Negative - плохие клиенты, которых машина назвала хорошими
FP = 23
Accuracy = (TP + TN) / (TP + TN + FP + FN)
Accuracy
```

$$Recall = \frac{TP}{TP+FP}$$
Доля обнаруженных плохих клиентов

$$Precision = \frac{TP}{TP+FN}$$
Доля действительно плохих клиентов среди тех, кого машина назвала плохими

$$F1 = \frac{2PrecisionRecall}{Precision + Recall}$$
Среднее гармоническое Precision и Recall

```python
Precision, Recall = TP / (TP + FP), TP / (TP + FN)
F1 = 2 * Precision * Recall / (Precision + Recall)
Precision, Recall, F1
```

```python
accuracy_score(y_train, logreg.predict(x_train)), precision_score(y_train, logreg.predict(x_train)), recall_score(y_train, logreg.predict(x_train)), f1_score(y_train, logreg.predict(x_train))
```
